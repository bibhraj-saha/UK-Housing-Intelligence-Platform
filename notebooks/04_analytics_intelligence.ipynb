# Affordability Analytics

The objective of affordability analytics is to identify areas where property prices are relatively affordable compared to other areas.

Lower property prices should result in higher affordability scores.

The initial implementation uses min-max normalization on median property prices.

## Affordability Score Version 1 (Min-Max Normalization)

Method:

1. Aggregate property transactions to area level (LSOA).
2. Calculate median property price for each area.
3. Apply min-max normalization.
4. Reverse the score so that lower prices receive higher scores.

Formula:

Affordability Score = 100 − Normalized Median Price

Expected Outcome:

- Cheapest area → Score near 100
- Most expensive area → Score near 0

Validation is required to determine whether the resulting score distribution is suitable for ranking purposes.

```python
area_df["affordability_score"] = (
    100 -
    (
        (
            area_df["median_price"]
            -
            area_df["median_price"].min()
        )
        /
        (
            area_df["median_price"].max()
            -
            area_df["median_price"].min()
        )
    ) * 100
)

area_df[
    [
        "lsoa_code",
        "median_price",
        "affordability_score"
    ]
].head()
```

## Validation Findings

The min-max affordability score produced a highly compressed distribution.

Observations:

- Mean score ≈ 93
- Most areas received scores between 90 and 100
- A small number of extremely expensive areas distorted the normalization range

Conclusion:

The score is mathematically correct but not suitable for ranking because it does not sufficiently differentiate between most areas.

## Affordability Score Version 2 (Percentile-Based)

To reduce the impact of extreme property price outliers, percentile ranking is used instead of direct min-max normalization.

Method:

1. Rank all areas by median property price.
2. Convert ranks into percentiles.
3. Reverse the percentile score so that lower-priced areas receive higher scores.

Advantages:

- Less sensitive to extreme values
- Produces a more balanced score distribution
- Better supports ranking and comparison between areas

```python
area_df["affordability_percentile"] = (
    area_df["median_price"]
    .rank(pct=True)
)

area_df["affordability_score_v2"] = (
    100 -
    (area_df["affordability_percentile"] * 100)
)

area_df["affordability_score_v2"].describe()
```

## Final Affordability Scoring Decision

Percentile-based affordability scoring has been adopted as the production methodology.

Results:

- Mean score ≈ 50
- Balanced distribution across all areas
- Improved differentiation between neighbourhoods
- More suitable for area ranking and composite index calculations

The percentile-based affordability score will be used in all subsequent analytics layers, including:

- Growth Score
- Investment Score
- Housing Intelligence Index
- Area Ranking Engine

# Crime Analytics

The objective of crime analytics is to identify safer areas based on recorded crime activity.

Lower crime levels should result in higher crime scores.

The initial implementation uses min-max normalization.

## Crime Score Version 1 (Min-Max Normalization)

Method:

1. Aggregate crime values to area level.
2. Calculate average crime per area.
3. Apply min-max normalization.
4. Reverse the score so that lower crime areas receive higher scores.

Formula:

Crime Score = 100 − Normalized Crime Rate

```python
area_df["crime_score_v2"] = (
    100 -
    (
        (
            area_df["average_crime"]
            -
            area_df["average_crime"].min()
        )
        /
        (
            area_df["average_crime"].max()
            -
            area_df["average_crime"].min()
        )
    ) * 100
)

area_df[
    [
        "lsoa_code",
        "average_crime",
        "crime_score_v2"
    ]
].head()
```

```python
area_df["crime_score_v2"].describe()
```

```python
area_df[
    [
        "lsoa_code",
        "average_crime",
        "crime_score_v2"
    ]
].sort_values(
    by="crime_score_v2",
    ascending=False
).head(10)
```

```python
area_df[
    [
        "lsoa_code",
        "average_crime",
        "crime_score_v2"
    ]
].sort_values(
    by="crime_score_v2",
    ascending=True
).head(10)
```

## Validation Findings

The min-max crime score showed severe compression.

Observations:

- Mean score ≈ 98
- Most areas received scores near 100
- A small number of high-crime outliers dominated the normalization range

Conclusion:

The score was not suitable for ranking because it failed to sufficiently separate most areas.

```python
area_df.to_parquet(
    "../data/analytics/crime_scores.parquet",
    index=False
)

print("Crime scores saved")
```

## Crime Score Version 2 (Percentile-Based)

A percentile-based approach was introduced to reduce the influence of extreme crime outliers.

Method:

1. Rank areas by crime level.
2. Convert ranks into percentiles.
3. Reverse the percentile score so that safer areas receive higher scores.

Advantages:

- More robust against outliers
- Balanced score distribution
- Better support for ranking applications

```python
area_df["crime_percentile"] = (
    area_df["average_crime"]
    .rank(pct=True)
)

area_df["crime_score_v3"] = (
    100 -
    (area_df["crime_percentile"] * 100)
)

area_df[
    [
        "average_crime",
        "crime_score_v2",
        "crime_score_v3"
    ]
].head()
```

```python
area_df["crime_score_v3"].describe()
```

## Final Crime Scoring Decision

Percentile-based crime scoring has been adopted as the production methodology.

Results:

- Mean score ≈ 50
- Balanced distribution
- Strong separation between safer and less safe areas
- Suitable for rankings and composite indexes

The percentile-based crime score will be used in:

- Growth Score calculations
- Investment Score calculations
- Housing Intelligence Index
- Area Ranking Engine

# Housing Intelligence Index (HII)

The Housing Intelligence Index is the master score of the UK Housing Intelligence Platform.

Purpose:

Provide a single overall measure of area attractiveness by combining multiple analytical dimensions.

Inputs:

- Affordability Score
- Crime Score
- Growth Score
- Investment Score

The index is designed to support:

- Area comparison
- Area ranking
- Investment discovery
- Dashboard visualisation

## Housing Intelligence Index Weighting

| Component | Weight |
|------------|----------|
| Affordability Score | 25% |
| Crime Score | 20% |
| Growth Score | 25% |
| Investment Score | 30% |

Total Weight = 100%

```python
area_df["housing_intelligence_index"] = (
    area_df["affordability_score"] * 0.25
    +
    area_df["crime_score"] * 0.20
    +
    area_df["growth_score"] * 0.25
    +
    area_df["investment_score"] * 0.30
)
```
